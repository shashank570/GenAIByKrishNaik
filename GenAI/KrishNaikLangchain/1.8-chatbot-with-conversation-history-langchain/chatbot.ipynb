```python
import os
from dotenv import load_dotenv
load_dotenv()
groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(model = "llama-3.1-8b-instant", groq_api_key = groq_api_key)
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content="Hi, my name is Shashank and I am a Lead Engineer in NTT Data")])
```

```python
from langchain_core.messages import HumanMessage, AIMessage
model.invoke(
    [
        HumanMessage(content="Hi, my name is Shashank and I am a Lead Engineer in NTT Data"),
        AIMessage(content="I don't have any information about your name. I'm a large language model, I don't have the ability to retain personal information or recall previous conversations. Each time you interact with me, it's a new conversation. If you'd like to share your name, I'd be happy to chat with you!"),
        HumanMessage(content="What is my name and what do I do?"),
    ]
)
```

```python
## Message History
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}

# if not exists create a new instance else return the existing one
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable" : {"session_id" : "chat1"}}
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="Hi, my name is Shashank and I am Lead Engineer in NTT Data")],
    config = config
)
response
```

```python
with_message_history.invoke(
    [HumanMessage(content="What is my name and what i do?")],
    config = config
)
```

```python
# Prompt Template
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder 
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant. Answer all questions to the best of your ability in {language}."),
        MessagesPlaceholder(variable_name="messages")
    ]
)
chain = prompt | model
```

```python
response = chain.invoke(
        {
            "messages" : [HumanMessage(content = "Hi my name is Shashank")],
            "language" : "Hindi"
        }
    )

response.content
```

```python
with_message_history=RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)

config = {"configurable" : {"session_id" : "chat3"}}
with_message_history.invoke(
    {
        "messages" : [HumanMessage(content="Hi, I am Shashank")],
        "language" : "Hindi"
    },
    config = config
)
```

### Managing the Conversation History
One important concept to understand when building chatbots is how to manage conversation history. If left unmanaged, the list of messages will grow unbounded and potentially overflow the context window of the LLM. Therefore, it is important to add a step that limits the size of the messages you are passing in.
'trim_messages' helper to reduce how many messages we're sending to the model. The trimmer allows us to specify how many tokens we want to keep, along with other parameters like if we want to always keep the system message and whether to allow partial messages

```python
from langchain_core.messages import SystemMessage, trim_messages
trimmer = trim_messages(
    max_tokens = 55, ## takes only 45 tokens in the message list
    strategy = "last", ## takes the most recent message
    token_counter = model, ## model is used to count tokens
    include_system = True, # always include the system token
    allow_partial = False, 
    start_on = "human" ## always start with human message
)

messages = [
    SystemMessage(content="you're a good assistant"),
    HumanMessage(content="hi! I'm bob"),
    AIMessage(content="hi!"),
    HumanMessage(content="I like vanilla ice cream"),
    AIMessage(content="nice"),
    HumanMessage(content="whats 2 + 2"),
    AIMessage(content="4"),
    HumanMessage(content="thanks"),
    AIMessage(content="no problem!"),
    HumanMessage(content="having fun?"),
    AIMessage(content="yes!"),
]

trimmer.invoke(messages)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough

chain = (
    RunnablePassthrough.assign(messages=itemgetter("messages") | trimmer) 
    | prompt | model
)

response = chain.invoke(
    {
        "messages" : messages + [HumanMessage(content="What ice creame do i like?")],
        "language" : "English"
    }
)
response.content
```

```python
response = chain.invoke(
    {
        "messages": messages + [HumanMessage(content="what math problem did i ask")],
        "language": "English",
    }
)
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key = "messages"
)
config = {"configurable" : {"session_id" : "chat4"}}
```

```python
response = with_message_history.invoke(
    {
        "messages" : messages + [HumanMessage(content="What is my name?")],
        "language" : "Hindi"
    },
    config = config
)
response.content
```
