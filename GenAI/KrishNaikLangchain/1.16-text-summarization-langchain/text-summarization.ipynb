## Prompt Template Text Summarization

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

generictemplate = """
Write the summary of the following speech
Speech: {speech}
Translate the precise summary to {language}
"""
```

```python
prompt = PromptTemplate(
    input_variables = ["speech", "language"],
    template = generictemplate
)

stroutputparser = StrOutputParser()

chain = prompt | llm | stroutputparser

response = chain.invoke(
    {
        "speech" : speech,
        "language" : "Hindi"
    }
)

print(response)
```

## Text Summarization

In modern versions of LangChain Docs, many legacy chains are being replaced by LCEL (LangChain Expression Language) using RunnableSequence.

StuffDocumentChain Text summarization

- Combines all documents into one prompt
- Sends everything to the LLM at once
- Good for small text

map_reduce text summarization

- Summarize each chunk separately (map)
- Combine summaries into final summary (reduce)
- Chunk1 → Summary1 -> Chunk2 → Summary2 -> Chunk3 → Summary3 -> Combine → Final Summary


refine -> Iterative summarization.

- Create initial summary from first chunk
- Refine summary using next chunk
- Continue refining
- Flow -> Chunk1 → Initial Summary + Chunk2 → Refined Summary + Chunk3 → Refined Again

```python
## Stuff Document Chain
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader("apjspeech.pdf")
documents = loader.load()

print(f"Total Pages Loaded: {len(documents)}")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=200
)

docs = text_splitter.split_documents(documents)
print(f"Total Chunks: {len(docs)}")
```

```python
output_parser = StrOutputParser()

print("\n==============================")
print("RUNNING STUFF CHAIN")
print("==============================\n")

# Combine all chunks into one text
full_text = "\n".join([doc.page_content for doc in docs])
full_text
```

```python
stuff_template = """
    Write a detailed summary of the following text.
    TEXT:
    {text}
    SUMMARY:
"""

stuff_prompt = PromptTemplate(
    input_variables=["text"],
    template=stuff_template
)

stuff_chain = stuff_prompt | llm | output_parser

stuff_summary = stuff_chain.invoke({
    "text": full_text
})

print("\nSTUFF SUMMARY:\n")
print(stuff_summary)
```

```python
# ==========================================
# ==========================================
# 2. MAP REDUCE CHAIN
# ==========================================
# ==========================================
```

```python
print("\n==============================")
print("RUNNING MAP REDUCE CHAIN")
print("==============================\n")
```

```python

map_template="""
    Summarize the following chunk of text.
    TEXT:
    {text}
    SUMMARY:
"""

map_prompt = PromptTemplate(
    input_variables = ["text"],
    template = map_template
)

map_chain = map_prompt | llm | output_parser
```

```python
# MAP STEP

chunk_summaries = []

for i, doc in enumerate(docs):
    print(f"Processing Chunk {i+1}/{len(docs)}")
    summary = map_chain.invoke({
        "text": doc.page_content
    })
    chunk_summaries.append(summary)
```

```python
# REDUCE STEP
combined_summaries = "\n".join(chunk_summaries)

reduce_prompt = PromptTemplate(
    input_variables=["text"],
    template="""
    Combine the following summaries into one final detailed summary.

    TEXT:
    {text}

    FINAL SUMMARY:
    """
)

reduce_chain = reduce_prompt | llm | output_parser

final_map_reduce_summary = reduce_chain.invoke({
    "text": combined_summaries
})

print("\nMAP REDUCE SUMMARY:\n")
print(final_map_reduce_summary)
```

```python
# ==========================================
# ==========================================
# 3. REFINE CHAIN
# ==========================================
# ==========================================
```

```python
print("\n==============================")
print("RUNNING REFINE CHAIN")
print("==============================\n")
```

```python
# INITIAL SUMMARY PROMPT

initial_refine_template = """
    Write an initial summary of the following text.
    TEXT:
    {text}
    SUMMARY:
"""

initial_refine_prompt = PromptTemplate(
    input_variables=["text"],
    template = initial_refine_template
)

initial_refine_chain = initial_refine_prompt | llm | output_parser

final_refine_template = """
    Existing Summary:
    {existing_summary}
    New Context:
    {text}
    Refine the existing summary using the new context.
"""

final_refine_prompt = PromptTemplate(
    input_variables=["existing_summary", "text"],
    template = final_refine_template
)

final_refine_chain = final_refine_prompt | llm | output_parser

# INITIAL SUMMARY

refined_summary = initial_refine_chain.invoke({
    "text": docs[0].page_content
})

# ITERATIVE REFINEMENT

for i, doc in enumerate(docs[1:]):
    print(f"Refining with Chunk {i+2}/{len(docs)}")

    refined_summary = final_refine_chain.invoke({
        "existing_summary": refined_summary,
        "text": doc.page_content
    })

print("\nREFINE SUMMARY:\n")
print(refined_summary)
```
