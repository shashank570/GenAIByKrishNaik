```python
import ollama
response = ollama.chat(
    model = "gemma:2b",
    messages = [
        {"role" : "system", "content" : "You are helpful assistant"},
        {"role" : "user", "content" : "What is Deep Learning"},
    ]
)
print(response["message"]["content"])
```

```python
import ollama
stream = ollama.chat(
    model = "gemma:2b",
    messages = [
        {"role" : "system", "content" : "You are helpful assistant"},
        {"role" : "user", "content" : "What is Deep Learning"},
    ],
    stream = True
)

for chunk in stream:
    print(chunk["message"]["content"], end="", flush=True) 
    # end = "" -> prevent new line after each chunk
    # flush = True -> Forces Python to immediately print output
```
