```python
# !pip install tansformers -q
from transformers import AutoTokenizer
```

```python
# loading a pre-trained Tokenizer
tokenizer = AutoTokenizer.from_pretrained("google-bert/bert-base-uncased")
```

```python
text = "I love Machine Learning!"
tokens = tokenizer.tokenize(text)

print("Tokens:", tokens)
```

```python
# tokens to token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("Token IDs:", token_ids)
```

```python
text = "I love Machine Learning!"
encoded = tokenizer(text)
print(encoded)
```

```python
# [Cls] -> 101
# [Sep] -> 102

# decode back (after a model generates output as token ids)
decoded = tokenizer.decode([101, 1045, 2293, 3698, 4083, 999, 102])

print("Decoded Text:", decoded)
```

```python
# subword behaviour
text = "unhappiness"
tokens = tokenizer.tokenize(text)
print(tokens)
# ## means → continuation of previous word
```
