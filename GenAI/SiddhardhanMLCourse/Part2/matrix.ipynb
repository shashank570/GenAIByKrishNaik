Scaler -> Scaler is a single number without direction
Vectors -> Vectors is a list of number. it can have only single row or only single column
Matrix -> Matrix can have multiple rows and multiple columns

```python
import numpy as np
```

```python
matrix_1 = np.asarray([[2, 3],[6, 7]])
print(matrix_1)
```

```python
matrix_2 = np.asarray([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
print(matrix_2)
```

```python
# matrix with random values
random_matrix = np.random.randint(100, size = (4,5))
print(random_matrix)
```

```python
# Matrix with all the values as 1
matrix_3 = np.ones((2, 3))
print(matrix_3)
```

```python
matrix_3 = np.ones((10, 10), dtype=int)
print(matrix_3)
```

```python
null_matrix = np.zeros((4, 4), dtype=int)
print(null_matrix)
```

```python
# Identity Matrix
identity_matrix = np.eye(5,5, dtype=int)
print(identity_matrix)
```

```python
# Transpose of a Matrix
a = np.random.randint(100, size=(4, 5))
print(a)
```

```python
transpose_of_a = np.transpose(a)
print(transpose_of_a)
```

```python
a = np.asarray([[2, 3], [4, 5]])
b = np.asarray([[4, 5], [6, 7]])
print(a)
print(b)
print('sum = ', a+b)
```

```python
a = np.asarray([[2, 3], [4, 5]])
b = np.asarray([[4, 5], [6, 7]])
print(a)
print(b)
print('diff = ', a-b)
```

```python
# we can only add two matrix if they are of same shape
# operands could not be broadcast together with shapes (3,3) (3,4) 
matrix_1 = np.random.randint(10, size=(3,3))
matrix_2 = np.random.randint(10, size=(3,3))
sum = np.add(matrix_1, matrix_2)
print(matrix_1)
print(matrix_2)
print(sum)
```

```python
diff = np.subtract(matrix_1, matrix_2)
print(diff)
```

```python
# multiplication with scaler 
x = 5
y = np.random.randint(10, size=(4,4))
product = np.multiply(x, y)
print(y)
print(product)
```

```python
# Multiplying 2 Matrices
matrix_1 = np.random.randint(5, size=(3, 3))
matrix_2 = np.random.randint(5, size=(3, 4))
print(matrix_1)
```

```python
print(matrix_2)
```

```python
product = np.dot(matrix_1, matrix_2)
print(product)
```

```python
# Element wise Matrix Multiplication -> both matrix should have same shape
# create two matrices with random values
matrix_3 = np.random.randint(5, size=(3,3))
matrix_4 = np.random.randint(5, size=(3,3))
```

```python
print(matrix_3)
```

```python
print(matrix_4)
```

```python
product = np.multiply(matrix_3, matrix_4)
print(product)
```
