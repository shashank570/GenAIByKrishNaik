```python
# dot product -> A · B = |A| |B| cos(θ) -> resultant would be a scaler
# cross product -> |A × B| = |A| |B| sin(θ) -> resultant would be a vector

a = np.asarray([2, 3])
b = np.asarray([4, 4])

a_dot_b = np.dot(a, b)
print(a_dot_b)
```

```python
c = np.asarray([40, 20, 35])
d = np.asarray([53, 24, 68])
c_dot_d = np.dot(c, d)
print(c_dot_d)
```

```python
# cross product of 2 vector
a = np.asarray([2, 3])
b = np.asarray([4, 4])

a_cross_b = np.cross(a, b)
print(a_cross_b)
```

```python
c = np.asarray([5, 10, 20])
d = np.asarray([18, 32, 50])
c_cross_d = np.cross(c, d)
print(c_cross_d)
```

```python
a = np.array([2, 5])

v = np.array([8, -6])

# magnitude of "v" vector
magnitude_of_v = np.sqrt(np.sum(v**2))

proj_of_a_on_v = (np.dot(a,v)/magnitude_of_v**2)*v 

print('Projection of a vector on v vector = ', proj_of_a_on_v)
```

```python
a = np.asarray([23, 45, 62])
v = np.asarray([45, 82, 67])

magnitude_of_v = np.sqrt(np.sum(v**2)) 

proj_of_a_on_v = (np.dot(a, v)/magnitude_of_v**2)*v

print('Projection of a vector on v vector = ', proj_of_a_on_v)
```

Scaler -> Scaler is a single number without direction
Vectors -> Vectors is a list of number. it can have only single row or only single column
