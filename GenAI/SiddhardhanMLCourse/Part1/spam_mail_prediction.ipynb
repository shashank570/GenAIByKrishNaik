```python
import re
import numpy as np
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
raw_mail_data = pd.read_csv('mail_data.csv')
```

```python
raw_mail_data.head()
```

```python
# replace the null values with a null string
# pd.notnull(raw_mail_data) -> Checks each value in dataset -> True  → value exists -> False → value is missing (NaN)
# .where(condition, value_if_false) -> If condition is True → keep original value -> If condition is False → replace with ''
mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)), '')
```

```python
mail_data.head()
```

```python
mail_data.shape
```

```python
# finding the count of different category
mail_data['Category'].value_counts()
```

```python
# load the Label Encoder function
label_encoder = LabelEncoder()
```

```python
labels = label_encoder.fit_transform(mail_data.Category)
```

```python
mail_data['Target'] = labels
```

```python
X = mail_data['Message']
Y = mail_data['Target']
```

```python
import nltk
nltk.download("english")
```

```python
print(stopwords.words('english'))
```

```python
port_stem = PorterStemmer()
```

```python
def stemming(content: str):
    stem_content  = re.sub('[^a-zA-Z]', ' ', content)
    stem_content = stem_content.lower()
    stem_content = stem_content.split()
    stem_content = [port_stem.stem(word) for word in stem_content if not word in stopwords.words('english')]
    stem_content = " ".join(stem_content)
    return stem_content
    
```

```python
mail_data['Message'] = mail_data['Message'].apply(stemming)
```

```python
X = mail_data['Message']
```

```python
print(X)
```

```python
# Splitting the data into training data & test data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 3)
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

```python
# transform the text data to feature vectors that can be used as input to the Logistic regression
feature_extraction = TfidfVectorizer()

X_train_features = feature_extraction.fit_transform(X_train)
X_test_featrures = feature_extraction.transform(X_test)

Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

```python
print(X_train)
```

```python
print(X_train_features)
```

```python
# Logistic Regression
model = LogisticRegression()
```

```python
# training the Logistic Regression model with the training data
model.fit(X_train_features, Y_train)
```

```python
prediction_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(prediction_on_training_data, Y_train)
```

```python
print('Accuracy on training data : ', accuracy_on_training_data)
```

```python
prediction_on_test_data = model.predict(X_test_featrures)
accuracy_on_testing_data = accuracy_score(prediction_on_test_data, Y_test)
```

```python
print('Accuracy on testing data : ', accuracy_on_testing_data)
```
