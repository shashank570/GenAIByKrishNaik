```python
from langchain_community.document_loaders import DirectoryLoader, UnstructuredFileLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
!pwd
```

```python
# configuration
docs_dir_path = "/Users/shashankshukla/Downloads/GenAIByKrishNaik/GenAI/SiddhardhanMLCourse/RAG/docs_dir"
vector_db_path = "/Users/shashankshukla/Downloads/GenAIByKrishNaik/GenAI/SiddhardhanMLCourse/RAG/vector_db"
collection_name = "document_collection"
```

```python
# loading the embedding model
embedding = HuggingFaceEmbeddings()
```

```python
# directory loader
loader = DirectoryLoader(
    path = docs_dir_path,
    glob = "./*.pdf", # File filter pattern
    loader_cls = UnstructuredFileLoader # Specifies how each file is read -> Extracts text from PDFs, Word, HTML, etc.
)
```

```python
documents = loader.load()
```

```python
print(type(documents))
```

```python
len(documents)
```

```python
# initializing the text splitter
text_splitter = CharacterTextSplitter(
    chunk_size = 2000,
    chunk_overlap = 500
)
```

```python
# splitting the text into smaller chunks
text_chunks = text_splitter.split_documents(documents)
```

```python
len(text_chunks)
```

```python
text_chunks[0]
```

```python
# creating the vector store
vector_store = Chroma.from_documents(
    documents = text_chunks,
    embedding = embedding,
    persist_directory = vector_db_path, # Where data is stored on disk
    collection_name = collection_name # Logical name for your dataset -> think it as a table name in a database
)
```
