```python
# configuration
docs_dir_path = "/Users/shashankshukla/Downloads/GenAIByKrishNaik/GenAI/SiddhardhanMLCourse/RAG/docs_dir"
vector_db_path = "/Users/shashankshukla/Downloads/GenAIByKrishNaik/GenAI/SiddhardhanMLCourse/RAG/llama_vector_db"
collection_name = "documents_collection"
```

```python
# Document Ingestion
# Define embedding Function
embed_model = HuggingFaceEmbedding() 
```

```python
# directory loader
loader = SimpleDirectoryReader(input_dir  = docs_dir_path)
```

```python
# load the documents
documents = loader.load_data()
```

```python
print(len(documents))
```

```python
# Define persistent DB loaction
db = chromadb.PersistentClient(path=vector_db_path)
```

```python
# Create or retrieve the vector collection
chroma_collection = db.get_or_create_collection(name=collection_name)
```

```python
# Create parser with chunking strategy
parser = SimpleNodeParser.from_defaults(chunk_size = 1024, chunk_overlap = 50)
```

```python
# Convert documents to chunks (nodes)
nodes = parser.get_nodes_from_documents(documents)
```

```python
# Create vector store
vector_store = ChromaVectorStore(chroma_collection=chroma_collection)
```

```python
# Create storage context
storage_context  = StorageContext.from_defaults(vector_store = vector_store)
```

```python
# Create the vector store index
index = VectorStoreIndex(
    nodes,
    storage_context = storage_context,
    vector_store = vector_store,
    embed_model = embed_model
)
```
