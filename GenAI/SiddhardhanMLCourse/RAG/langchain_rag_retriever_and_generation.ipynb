```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
from langchain_classic.chains import RetrievalQA
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
```

```python
vector_db_path = "/Users/shashankshukla/Downloads/GenAIByKrishNaik/GenAI/SiddhardhanMLCourse/RAG/vector_db"
collection_name = "document_collection"
```

```python
embedding = HuggingFaceEmbeddings()
```

```python
llm = ChatGroq(
    model = "llama-3.1-8b-instant",
    temperature = 0.0
)
```

```python
vector_store = Chroma(
    collection_name = collection_name,
    embedding_function = embedding,
    persist_directory = vector_db_path
)
```

```python
retriever = vector_store.as_retriever()
```

```python
qa_chain = RetrievalQA.from_chain_type(
    llm = llm,
    chain_type = "stuff",
    retriever = retriever,
    return_source_documents = True
)

# "stuff" means:
# Take all retrieved docs
# “Stuff” them into one prompt

# Output will include:
# Answer
# Source documents used
```

```python
query = "What does the document say about Adaptive Radiation?"
response = qa_chain.invoke({"query" : query})
print(response["result"])
print("="*100)
for source in response["source_documents"]:
    print(source.metadata)
```
