```python
from ollama import Client
```

```python
client = Client(
    host = "http://localhost:11434" # when we run in ec2 we provide ec2 public id
)

response = client.chat(
    model = "gemma2:2b",
    messages = [
        {"role" : "system", "content" : "You are a helpful chat assistant"},
        {"role" : "user", "content" : "What is Machine learning in 20 words?"}
    ],
    stream = True
)

for chunk in response:
    print(chunk["message"]["content"], end = "", flush = True)
```
